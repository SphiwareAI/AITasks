# **Sentece 1**

```python
# levels
# 1: who ?
# 2: doing what ?
# 3: where ?
# a young(age) boy(Person, male) is playing(action, this is more specific) football(object) in the park(location, outdoors)

# step 1
levels_1 = {
    1: {
        "person": "boy",
        "action": "playing",
    },
    2: {
        "object": "football",
        "setting": "outdoors",
    },
    3: {
        "age": "young",
        "location": "park"
    }
}

import pandas as pd
levels_dataframe_1 = pd.DataFrame.from_dict(levels_1, orient='index')
print("Printing levels_1 as a dataframe")
print("="*100)

print(levels_1)
print("="*100)
print(levels_dataframe_1)

representations_1 = {}
for level_number in sorted(levels_1):
    representations_1.update(levels_1[level_number])
    print(f"\n After Level {level_number}:")
    print(representations_1)



# now turn this information into a sentence: 

sentence_1 = (
    f"A {representations_1['age']} "
    f"{representations_1['person']} is "
    f"{representations_1['action']} "
    f"{representations_1['object']} in the "
    f"{representations_1['location']}."

)
    

print("Print reconstruced sentence")
print("="*100)
print(sentence_1)
```

# **Sentence 2**

```python
# Sentence 2:
# A bakkie is parked inside the garage

# Levels:
# Level 1: What is the main object? What is its state?
# Level 2: What is its general setting?
# Level 3: What is its precise spatial relationship?

# A bakkie (vehicle)
# is parked (state)
# inside (spatial relationship)
# the garage (location, indoors)

levels_2 = {
    1: {
        "object": "bakkie",
        "state": "parked",
    },
    2: {
        "setting": "indoors",
        "location": "garage",
    },
    3: {
        "spatial_relation": "inside",
    }
}


levels_dataframe_2 = pd.DataFrame.from_dict(
    levels_2,
    orient="index"
)

print("Printing levels_2 as a dataframe")
print("=" * 100)

print(levels_2)
print("=" * 100)

print(levels_dataframe_2)


representations_2 = {}

for level_number in sorted(levels_2):
    representations_2.update(levels_2[level_number])

    print(f"\nAfter Level {level_number}:")
    print(representations_2)


sentence_2 = (
    f"A {representations_2['object']} is "
    f"{representations_2['state']} "
    f"{representations_2['spatial_relation']} "
    f"the {representations_2['location']}."
)

print("Reconstructed sentence")
print("=" * 100)
print(sentence_2)
```

# **Sentence 3**

```python
# Sentence 3:
# A bottle of Coca-Cola is placed on top of the table

# Levels:
# Level 1: What is the main object? What is happening to it?
# Level 2: What is the general setting?
# Level 3: What is the spatial relationship and what additional details
#          identify the object?

# A bottle (object)
# of Coca-Cola (contents or brand)
# is placed (action/state)
# on top of (spatial relationship)
# the table (location, indoors)

levels_3 = {
    1: {
        "object": "bottle",
        "action": "placed",
    },
    2: {
        "setting": "indoors",
        "location": "table",
    },
    3: {
        "spatial_relation": "on top of",
        "contents_or_brand": "Coca-Cola",
    }
}

levels_dataframe_3 = pd.DataFrame.from_dict(
    levels_3,
    orient="index"
)

print("Printing levels_3 as a dataframe")
print("=" * 100)

print(levels_3)
print("=" * 100)

print(levels_dataframe_3)

representations_3 = {}

for level_number in sorted(levels_3):
    representations_3.update(levels_3[level_number])

    print(f"\nAfter Level {level_number}:")
    print(representations_3)

sentence_3 = (
    f"A {representations_3['object']} "
    f"of {representations_3['contents_or_brand']} is "
    f"{representations_3['action']} "
    f"{representations_3['spatial_relation']} "
    f"the {representations_3['location']}."
)

print("Reconstructed sentence")
print("=" * 100)
print(sentence_3)
```

```python
import subprocess


import json
import requests

result = subprocess.run(
    ["ollama", "list"],
    capture_output=True,
    text=True
)

print(result.stdout)


OLLAMA_URL = "http://localhost:11434"

response = requests.get(
    f"{OLLAMA_URL}/api/tags",
    timeout=10
)

print("Status code:", response.status_code)
print(response.json())
```

```python
import json

model = "qwen3:8b"

sentence = "A young boy is playing football in the park"

prompt = """
Represent the sentence below using exactly three semantic levels.

LEVEL 1:
Identify the main entity and its main action or state.

LEVEL 2:
Identify the main object involved and the general setting,
where applicable.

LEVEL 3:
Identify additional attributes and specific contextual details,
such as age, size, precise location, spatial relationships, or time.

Rules:
- Preserve the meaning of the original sentence.
- Do not invent information that is not stated or clearly implied.
- Use concise, consistent semantic labels.
- Every level must be a JSON object.
- Return valid JSON only, without markdown or explanations.

Example output for the sentence:
"A young boy is playing football in the park"

{
    "1": {
        "person": "boy",
        "action": "playing"
    },
    "2": {
        "object": "football",
        "setting": "outdoors"
    },
    "3": {
        "age": "young",
        "location": "park"
    }
}

Sentence to represent:
""" + sentence
```

```python
response = requests.post(
    f"{OLLAMA_URL}/api/generate",
    json={
        "model": model,
        "prompt": prompt,
        "stream": False,
        "format": "json",
        "options": {
            "temperature": 0
        }
    },
    timeout=180
)

response.raise_for_status()

result = response.json()

print(result["response"])

levels_1 = json.loads(result["response"])

print(levels_1)
```

```python
import requests
import json

model = "qwen3:8b"

original_sentences = [
    "A young boy is playing football in the park",
    "A bakkie is parked inside the garage",
    "A bottle of cocacola is placed on top of the table",
    "A small refrigerator is switched off today"
]

prompt = """
You are generating sentences for a semantic representation experiment.

For EACH of the four original sentences below, generate EXACTLY
THREE new sentences inspired by that sentence.

Original sentences:

1. A young boy is playing football in the park
2. A bakkie is parked inside the garage
3. A bottle of cocacola is placed on top of the table
4. A small refrigerator is switched off today

Requirements:
- Generate 3 new sentences for each original sentence.
- Generate exactly 12 new sentences in total.
- Keep the sentences simple and grammatically correct.
- Preserve the general structure and meaning pattern of each original.
- Change relevant entities, actions, attributes, locations,
  spatial relationships, or time details.
- Do not copy an original sentence unchanged.
- Keep each generated sentence associated with its original sentence.
- Return valid JSON only, without markdown.

Use this structure:

{
    "original_1": {
        "original_sentence": "A young boy is playing football in the park",
        "generated_sentences": [
            "Generated sentence 1",
            "Generated sentence 2",
            "Generated sentence 3"
        ]
    },
    "original_2": {
        "original_sentence": "A bakkie is parked inside the garage",
        "generated_sentences": [
            "Generated sentence 1",
            "Generated sentence 2",
            "Generated sentence 3"
        ]
    },
    "original_3": {
        "original_sentence": "A bottle of cocacola is placed on top of the table",
        "generated_sentences": [
            "Generated sentence 1",
            "Generated sentence 2",
            "Generated sentence 3"
        ]
    },
    "original_4": {
        "original_sentence": "A small refrigerator is switched off today",
        "generated_sentences": [
            "Generated sentence 1",
            "Generated sentence 2",
            "Generated sentence 3"
        ]
    }
}
"""

response = requests.post(
    "http://localhost:11434/api/generate",
    json={
        "model": model,
        "prompt": prompt,
        "stream": False,
        "format": "json",
        "options": {
            "temperature": 0.7
        }
    },
    timeout=180
)

response.raise_for_status()

result = response.json()

print(result["response"])
```

```python
generated_sentences = json.loads(result["response"])

generated_sentences
```

```python
for original_key, data in generated_sentences.items():
    print(f"\n{original_key}")
    print("-" * 80)

    if isinstance(data, dict):
        original = data.get(
            "original_sentence",
            data.get("original", "Original sentence not provided")
        )

        sentences = data.get(
            "generated_sentences",
            data.get("generated", [])
        )

        print(f"Original: {original}")

    elif isinstance(data, list):
        sentences = data
    else:
        print("Unexpected data format:", data)
        continue

    if isinstance(sentences, list):
        for i, sentence in enumerate(sentences, start=1):
            print(f"{i}. {sentence}")

        print("Number generated:", len(sentences))
    else:
        print("Generated content:", sentences)
```
